```python
import pandas as pd
import matplotlib.pyplot as plt

journeys = pd.read_csv('autonomous_vehicle_journeys.csv')
journeys.head()
```

```python
print(journeys.shape)

journeys.info()

print(journeys.isna().sum())

print(journeys['problem_occurred'].value_counts())
```

```python
journeys['journey_date'] = pd.to_datetime(
    journeys['journey_date']
)

assert journeys['journey_id'].is_unique
```

```python
vehicles = pd.read_csv('vehicle_sensor_profiles.csv')
roads = pd.read_csv('road_environment_factors.csv')
weather = pd.read_csv('weather_risk_factors.csv')
events = pd.read_csv('intervention_events.csv')
```

```python
data = (
    journeys
    .merge(vehicles, on='vehicle_profile_id', how='left')
    .merge(roads, on='road_segment_id', how='left',
           suffixes=('', '_road'))
    .merge(weather, on='weather_code', how='left')
)

print(data.shape)

print(
    data[['vehicle_model', 'junction_complexity', 'weather']]
    .isna()
    .sum()
)
```

```python
data['problem_flag'] = (
    data['problem_occurred'] == 'Yes'
).astype(int)
```

```python
weather_rates = (
    data.groupby('weather')['problem_flag']
    .mean()
    .mul(100)
    .sort_values()
)

print(weather_rates.round(1))
```

```python
weather_rates.plot(kind='bar')

plt.ylabel('Problem rate (%)')
plt.title('Problem Rate by Weather')
plt.tight_layout()
plt.show()
```

```python
data.boxplot(
    column='sensor_confidence_pct',
    by='problem_occurred'
)

plt.suptitle('')
plt.title('Sensor Confidence by Journey Outcome')
plt.ylabel('Sensor confidence (%)')
plt.show()
pivot = pd.pivot_table(
    data,
    values='problem_flag',
    index='weather',
    columns='traffic_level',
    aggfunc='mean'
) * 100

print(pivot.round(1))
```

```python
pivot = pd.pivot_table(
    data,
    values='problem_flag',
    index='weather',
    columns='traffic_level',
    aggfunc='mean'
) * 100

print(pivot.round(1))
```

```python
features = [
    'time_of_day',
    'road_type',
    'traffic_level',
    'road_surface',
    'visibility_km',
    'distance_km',
    'sensor_confidence_pct',
    'sensor_alert_count',
    'unexpected_obstacle'
]

X = pd.get_dummies(data[features], drop_first=True)
y = data['problem_flag']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    ConfusionMatrixDisplay
)

print('Accuracy:', accuracy_score(y_test, y_pred))
print('Precision:', precision_score(y_test, y_pred))
print('Recall:', recall_score(y_test, y_pred))
print('F1 score:', f1_score(y_test, y_pred))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.show()
```

```python
importance = pd.Series(
    model.coef_[0],
    index=X.columns
)

importance = importance.reindex(
    importance.abs().sort_values(ascending=False).index
)

print(importance.head(12))
```
